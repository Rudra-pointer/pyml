```python
num = int(input("Enter a number: "))

if num <= 1:
    print("Not Prime")
else:
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print("Prime Number")
    else:
        print("Not Prime")
```

```python
def factorial():

    n = int(input("Enter a number below 50: "))

    if n >= 50:
        print("Please enter number below 50")

    else:
        fact = 1

        for i in range(1, n+1):
            fact = fact * i

        print("Factorial =", fact)

factorial()
```

```python
import csv

count = {}

file = open("Dataset_Day1.csv", "r")
data = csv.reader(file)

next(data)

for row in data:
    city = row[0]

    if city not in count:
        count[city] = 1
    else:
        count[city] += 1

print(count)

file.close()
```

```python
import csv

cities = set()

file = open("Dataset_Day1.csv", "r")

data = csv.reader(file)

next(data)

for row in data:

    city = row[0]
    hospital = row[1]

    if "apollo" in hospital.lower():
        cities.add(city)

print("Cities with Apollo hospital:")
print(cities)

file.close()
```
