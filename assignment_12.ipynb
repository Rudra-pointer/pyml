```python
import pandas as pd

# Load dataset
df = pd.read_csv("Dataset_Day12.csv")

# Columns where 0 means missing value
cols = ["Glucose", "BloodPressure", "BMI", "DiabetesPedigreeFunction"]

# Replace 0 with median value
for col in cols:

    # Replace 0 with NaN
    df[col] = df[col].replace(0, pd.NA)

    # Fill missing values using median
    df[col] = df[col].fillna(df[col].median())


# Remove outliers using IQR method
for col in df.columns[:-1]:      # Skip Outcome column

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_rows = df[(df[col] < lower) | (df[col] > upper)]

    df = df.drop(outlier_rows.index)


print(df.shape)
print(df.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Input and output variables
X = df.drop("Outcome", axis=1)
y = df["Outcome"]

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=100
)
model = GaussianNB()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Print performance metrics
print("Accuracy =", accuracy_score(y_test, y_pred))
print("Precision =", precision_score(y_test, y_pred))
print("Recall =", recall_score(y_test, y_pred))
print("F1 Score =", f1_score(y_test, y_pred))
```

```python
from sklearn.model_selection import cross_val_score

# Create model
model = GaussianNB()

# Perform 5-fold cross validation
scores = cross_val_score(
    model,
    X,
    y,
    cv=5,
    scoring="accuracy"
)

print("Accuracy of each fold:")
print(scores)

print("\nAverage Accuracy =", scores.mean())
```

```python
from sklearn.ensemble import AdaBoostClassifier

# Create Naive Bayes model
nb_model = GaussianNB()

# Create AdaBoost model
model = AdaBoostClassifier(
    estimator=nb_model,
    n_estimators=50,
    random_state=100
)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Performance metrics
print("Accuracy =", accuracy_score(y_test, y_pred))
print("Precision =", precision_score(y_test, y_pred))
print("Recall =", recall_score(y_test, y_pred))
print("F1 Score =", f1_score(y_test, y_pred))
```

```python
# Applying AdaBoost did not improve the model, 
# so the basic Naïve Bayes classifier remains the preferred choice for this dataset.
```
