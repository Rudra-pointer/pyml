```python
import pandas as pd

df = pd.read_csv("Dataset_Day4.csv")
print("Missing Values Before Treatment:")
print(df.isnull().sum())

#Treating missing values by filling with mean
df.TakeHome.fillna(df.TakeHome.mean(), inplace=True)
df.Final.fillna(df.Final.mean(), inplace=True)

print("\nMissing Values After Treatment:")
print(df.isnull().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Descriptive Statistics
print(df.describe())

# BOXPLOT
plt.figure(figsize=(10,6))
sns.boxplot(data=df)
plt.title("Box Plot")
plt.show()

# PAIRPLOT
sns.pairplot(df)
plt.show()


# HEATMAP
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(), annot=True)

plt.title("Correlation Heatmap")
plt.show()

# SCATTER PLOT
plt.figure(figsize=(10,6))

plt.scatter(df["Midterm"], df["Final"])

plt.title("MIDTERM VS FINAL")

plt.xlabel("Midterm")

plt.ylabel("Final")

plt.show()
```

```python
#1. Midterm and Final show a positive relationship.
#2. Students with higher Midterm marks generally have higher Final marks.
#3. The heatmap also shows positive correlation between Midterm and Final.
```
